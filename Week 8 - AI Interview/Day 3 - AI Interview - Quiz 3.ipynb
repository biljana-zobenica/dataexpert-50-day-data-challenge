## AI Interview - Quiz 3

### Scenario: Multi-modal AI Systems
### You are working on a project that combines visual and textual data (such as video with transcriptions) to generate insights.

### Question: Discuss the architecture for a multi-modal AI system that integrates these two data types. What challenges do you anticipate in feature extraction and fusion, and how would you evaluate model performance across different modalities?

Multi-Modal AI System Architecture

Use two separate parts:

- One model reads visual data (video frames or images)
- One model reads text (transcripts)

Each part turns its input into a numerical summary.

These summaries are then combined and passed to a final model that produces insights (classification, search, summaries).

Key Challenges

- Timing alignment: matching video moments with the right text
- Noisy data: transcripts may have errors; videos may be unclear
- Balance: making sure one data type doesn’t overpower the other

Evaluation

- Test the model using video only, text only, and both together
- Check if results improve when both data types are used
- Measure accuracy or quality depending on the task
