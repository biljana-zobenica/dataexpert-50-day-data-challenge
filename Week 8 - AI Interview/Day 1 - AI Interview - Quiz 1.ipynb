## AI Interview - Quiz 1

### Case: Zero-Shot Learning for Image Classification
### Imagine you are tasked with developing a system that can classify images into categories it has never seen before based on textual descriptions of those categories.

### Question: Explain the concept of zero-shot learning and how you would implement it in a deep learning framework. Discuss the model architectures and training techniques you'd use to achieve this.

What is Zero-Shot Learning (ZSL)?

Zero-shot learning enables a model to recognize classes it has never seen during training by leveraging side information (usually text) that describes those classes. Instead of learning a direct mapping from images → class IDs, the model learns a shared semantic space where images and text can be compared.

How to Implement Zero-Shot Learning:

- Core Idea

Represent images and class descriptions in the same embedding space
At inference, classify an image by finding the closest textual class embedding

Model Architecture:

- Image Encoder

CNN or Vision Transformer (e.g., ResNet, ViT)
Outputs an image embedding vector

- Text Encoder

Transformer-based language model (e.g., BERT, T5)
Encodes class descriptions (attributes, captions) into text embeddings

- Shared Embedding Space

Projection layers map image and text embeddings to the same dimension
Similarity computed via cosine similarity or dot product

This is the same principle used in CLIP-like models.

Training Technique:

- Contrastive Learning

Train on seen classes using paired image–text data:
Positive pairs: (image, correct description)
Negative pairs: (image, other descriptions)
Loss function:
Contrastive loss (InfoNCE / cross-entropy over similarities)

This teaches the model:
“Images of cats should be close to text describing cats, and far from dogs.”
