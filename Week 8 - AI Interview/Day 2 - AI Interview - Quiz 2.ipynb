## AI Interview - Quiz 2

### Scenario: Reinforcement Learning in Game Development
### Your team is building an AI agent to learn and adapt strategies in a complex game environment using reinforcement learning.

### Question: Describe the model architecture you would use. Discuss how you would manage exploration vs. exploitation, handle sparse rewards, and implement ongoing learning as the game evolves.

1. Model Architecture

- Core Setup

Agent–Environment loop following an MDP:

State → Action → Reward → Next State

- Model Choice

Deep Reinforcement Learning with neural networks:

Value-based: Deep Q-Network (DQN) for discrete action spaces

Policy-based: Policy Gradient / PPO for large or continuous action spaces

Actor–Critic (recommended):

Actor network outputs action probabilities

Critic network estimates value (V or Q)

- Inputs

Game state features (positions, resources, scores)

Optional CNN for visual input or LSTM for temporal context

2. Exploration vs. Exploitation

ε-greedy (for DQN): gradually decay ε to shift from exploration → exploitation

Entropy regularization (for PPO): encourages policy diversity

Noisy networks: learn exploration behavior directly

Curriculum learning: start with simpler opponents or scenarios

3. Handling Sparse Rewards

Reward shaping: add intermediate rewards for progress toward goals

Hierarchical RL: break tasks into sub-goals (options)

Intrinsic motivation:

Curiosity-based Rewards

Novelty or prediction error bonuses

Hindsight Experience Replay (HER) for goal-based tasks

4. Ongoing Learning & Game Evolution

Continuous training using self-play or simulated opponents

Experience replay buffers refreshed with recent gameplay

Periodic policy freezing to stabilize learning

Offline + online training mix to avoid catastrophic forgetting

5. Stability & Scaling Techniques

Target networks (for value stability)

Advantage normalization (Actor–Critic)

Clipping (PPO) to prevent destructive updates

Distributed rollouts for faster learning
