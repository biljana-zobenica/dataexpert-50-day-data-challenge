## Data Engineer Interview - Quiz 7

### Retail Sales Performance Analysis
### A retail client wants to track daily, weekly, and monthly sales performance by product, store, and region.

### Question: How would you design their sales warehouse schema, ensuring both detailed transaction tracking and aggregate reporting are supported?

Retail Sales Performance Analysis Solution by Dimensional Modeling (Warehouse Schema)

1. fact_sales

- sales_key (PK)
- date_key (FK → dim_date)
- product_key (FK → dim_product)
- store_key (FK → dim_store)
- units_sold
- sales_amount

Grain: one row per product–store–day transaction (or receipt line).

2. dim_product

- product_key (PK)
- product_name
- category
- brand

3. dim_store

- store_key (PK)
- store_name
- region_key (FK → dim_region)

4. dim_region

- region_key (PK)
- region_name

5. dim_date
- date_key (PK)
- day
- week
- month
- year

Aggregate Reporting

- Daily metrics come directly from fact_sales.
- Weekly and monthly reporting is derived by rolling up via dim_date.
- Optionally create aggregate tables (e.g., fact_sales_monthly) for high-volume environments. ✅
