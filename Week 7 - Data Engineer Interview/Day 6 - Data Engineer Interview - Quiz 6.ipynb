## Data Engineer Interview - Quiz 6

### Streaming App – Watch History
### A video platform wants to track what episodes users watch, when, and on which device. 
### **Question: **Propose a normalized schema. How would you extend this to track “likes” and “watch again” flags per episode?

Streaming App – Watch History Solution by Dimensional Modeling

1. fact_watch_events

- watch_event_key (PK)
- user_key (FK → dim_user)
- episode_key (FK → dim_episode)
- device_key (FK → dim_device)
- watch_date_key (FK → dim_date)
- watch_duration_minutes
- watch_count (always 1)

2. fact_episode_engagement

- engagement_key (PK)
- user_key (FK → dim_user)
- episode_key (FK → dim_episode)
- is_liked
- watch_again_flag
- updated_date_key (FK → dim_date)

3. dim_user

- user_key (PK)
- signup_date_key (FK → dim_date)

4. dim_episode

- episode_key (PK)
- series_name
- season_number
- episode_number
- duration_minutes

5. dim_device

- device_key (PK)
- device_type
- os
- dim_date
- date_key (PK)

Final points of data solution:
- Watch events captured at event grain for accurate viewing history
- Likes and “watch again” modeled separately as stateful engagement facts
- Star schema enables fast aggregation by user, episode, device, and time
- Append-only facts preserve full history for analytics ✅
